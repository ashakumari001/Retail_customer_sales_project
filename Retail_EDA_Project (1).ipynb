### 6.5 Checking date values

## 7. Data Cleaning

Cleaning decisions:
- Remove exact duplicate rows because they repeat the same transaction record.
- Standardize Gender and Region labels into one spelling.
- Convert mixed date formats into Python datetime values.
- Treat ages outside 0–100 and discounts outside 0–0.5 as invalid, then replace them with the median.
- Fill missing Age, Income and Discount values with their median because these are numerical columns and the median is less affected by extreme values.
- Keep missing Customer_Rating values as missing because a customer may not rate every order.
- Keep missing delivery dates missing instead of guessing a date.
- Replace missing Satisfaction_Level with `Not Rated`.
- If delivery date is earlier than order date, mark the delivery date as missing.

```python
clean = df.copy()

# Convert date columns. The source contains mixed date formats.
clean['Order_Date'] = pd.to_datetime(clean['Order_Date'], format='mixed', dayfirst=True, errors='coerce')
clean['Delivery_Date'] = pd.to_datetime(clean['Delivery_Date'], format='mixed', dayfirst=True, errors='coerce')

# Remove duplicate rows.
clean = clean.drop_duplicates().copy()

# Standardize Gender.
gender = clean['Gender'].astype('string').str.strip().str.lower()
clean['Gender'] = gender.map({
    'male': 'Male', 'm': 'Male',
    'female': 'Female', 'f': 'Female'
})

# Standardize Region.
region = clean['Region'].astype('string').str.strip().str.lower()
clean['Region'] = region.replace({
    'north': 'North', 'nort': 'North',
    'south': 'South', 'sth': 'South',
    'east': 'East',
    'west': 'West'
})

# Mark invalid numeric values as missing.
clean.loc[~clean['Age'].between(0, 100), 'Age'] = np.nan
clean.loc[~clean['Discount'].between(0, 0.5), 'Discount'] = np.nan

# Fill selected numeric missing values using median.
for col in ['Age', 'Income', 'Discount']:
    clean[col] = clean[col].fillna(clean[col].median())

# Keep ratings blank if no rating was provided.
clean['Satisfaction_Level'] = clean['Satisfaction_Level'].fillna('Not Rated')

# Invalid delivery sequence: do not keep an earlier delivery date.
bad_delivery = clean['Delivery_Date'] < clean['Order_Date']
clean.loc[bad_delivery, 'Delivery_Date'] = pd.NaT

# Drop any duplicate rows created after standardizing labels.
clean = clean.drop_duplicates().copy()

print("Rows before cleaning:", len(df))
print("Rows after cleaning:", len(clean))
print("Duplicates after cleaning:", clean.duplicated().sum())
display(clean.head())
```

## 8. Descriptive Statistics

```python
numeric_columns = [
    'Age', 'Income', 'Quantity', 'Unit_Price',
    'Discount', 'Sales', 'Cost', 'Profit'
]

summary = clean[numeric_columns].describe().T
summary['Median'] = clean[numeric_columns].median()
summary['Mode'] = clean[numeric_columns].mode().iloc[0]
summary['Variance'] = clean[numeric_columns].var()
summary['IQR'] = (
    clean[numeric_columns].quantile(0.75)
    - clean[numeric_columns].quantile(0.25)
)
display(summary.round(2))
```

## 9. Univariate Analysis

### 9.1 Numerical columns

```python
for col in ['Age', 'Income', 'Sales', 'Profit']:
    plt.figure(figsize=(7, 4))
    plt.hist(clean[col].dropna(), bins=30)
    plt.title('Distribution of ' + col)
    plt.xlabel(col)
    plt.ylabel('Frequency')
    plt.tight_layout()
    plt.show()

print("Skewness:")
display(clean[['Age', 'Income', 'Sales', 'Profit']].skew().to_frame('Skewness').round(2))
```

### 9.2 Categorical columns

```python
for col in ['Region', 'Product_Category', 'Customer_Segment']:
    print("\\nFrequency of", col)
    display(clean[col].value_counts().to_frame('Count'))
    clean[col].value_counts().plot(kind='bar', figsize=(7, 4), title='Count of ' + col)
    plt.ylabel('Number of transactions')
    plt.tight_layout()
    plt.show()
```

## 10. Outlier Detection Using IQR Method

```python
for col in ['Sales', 'Profit', 'Age', 'Income']:
    q1 = clean[col].quantile(0.25)
    q3 = clean[col].quantile(0.75)
    iqr = q3 - q1
    lower_limit = q1 - 1.5 * iqr
    upper_limit = q3 + 1.5 * iqr
    outliers = clean[(clean[col] < lower_limit) | (clean[col] > upper_limit)]

    print(col)
    print("Lower limit:", round(lower_limit, 2))
    print("Upper limit:", round(upper_limit, 2))
    print("Outlier count:", len(outliers))

    plt.figure(figsize=(7, 3))
    plt.boxplot(clean[col].dropna(), vert=False)
    plt.title('Outlier check - ' + col)
    plt.tight_layout()
    plt.show()
```

**Outlier interpretation:** IQR marks unusually low or high values; it does not automatically mean that a value is wrong. A high sales or income value may be a genuine large transaction or customer. Such values should be checked before deleting them. In this project, valid extreme sales and income values are retained.

## 11. Feature Engineering

```python
# 1. Profit margin as a percentage of sales.
clean['Profit_Margin'] = np.where(
    clean['Sales'] != 0,
    (clean['Profit'] / clean['Sales']) * 100,
    np.nan
)

# 2. Delivery time in days.
clean['Delivery_Days'] = (
    clean['Delivery_Date'] - clean['Order_Date']
).dt.days

# 3. Total revenue generated by each customer across all their orders.
clean['Revenue_per_Customer'] = clean.groupby('Customer_ID')['Sales'].transform('sum')

display(clean[
    ['Transaction_ID', 'Profit_Margin', 'Delivery_Days', 'Revenue_per_Customer']
].head())
```

## 12. Business Questions Using GroupBy

### Question 1: Which region has the highest total sales?

```python
region_sales = clean.groupby('Region')['Sales'].sum().sort_values(ascending=False)
display(region_sales.to_frame('Total Sales'))
```

### Question 2: Which product category gives the highest profit?

```python
category_profit = clean.groupby('Product_Category')['Profit'].sum().sort_values(ascending=False)
display(category_profit.to_frame('Total Profit'))
```

### Question 3: Which customer segment generates the highest profit?

```python
segment_profit = clean.groupby('Customer_Segment')['Profit'].sum().sort_values(ascending=False)
display(segment_profit.to_frame('Total Profit'))
```

### Question 4: Which payment method has the highest sales?

```python
payment_sales = clean.groupby('Payment_Method')['Sales'].sum().sort_values(ascending=False)
display(payment_sales.to_frame('Total Sales'))
```

### Question 5: Which delivery mode is used most often?

```python
delivery_orders = clean.groupby('Delivery_Mode').agg(
    Total_Orders=('Transaction_ID', 'count'),
    Average_Delivery_Days=('Delivery_Days', 'mean')
).sort_values('Total_Orders', ascending=False)
display(delivery_orders.round(2))
```

### Question 6: What is the average sales amount in each region?

```python
average_region_sales = clean.groupby('Region')['Sales'].mean().sort_values(ascending=False)
display(average_region_sales.to_frame('Average Sales').round(2))
```

## 13. Overall Charts

```python
plt.figure(figsize=(8, 4))
clean.groupby('Region')['Sales'].sum().sort_values().plot(kind='bar')
plt.title('Total Sales by Region')
plt.ylabel('Sales')
plt.tight_layout()
plt.show()

plt.figure(figsize=(8, 4))
clean.groupby('Product_Category')['Profit'].sum().sort_values().plot(kind='bar')
plt.title('Total Profit by Product Category')
plt.ylabel('Profit')
plt.tight_layout()
plt.show()
```
